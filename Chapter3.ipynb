**3.2 Experimental Studies (Limitations)**

```python
from time import time

def measure_time(algorithm, input_data):
    start_time = time()
    algorithm(input_data)
    end_time = time()
    return end_time - start_time

# LIMITATIONS of experimental approach:
# 1. Must implement algorithm first
# 2. Results may not generalize to other inputs
# 3. Need same hardware/software to compare algorithms
```

```python
Limitation | Explanation

Implementation required | Can't test before coding
Input dependent | Results vary with different data 
Envinronment dependent | Different computer give different times
```

**3.4 Pseudocode Conventions**

```python
Rules :
- Indentation replaces braces
- <- means assignment
- = means equality testing
```

```python
Algorithm find_max(data)
    Input: Array data of n numbers
    Output: Maximum element
    
    biggest ← data[0]
    for each val in data do
        if val > biggest then
            biggest ← val
        end if
    end for
    return biggest
```

**3.5 Seven Important Functions**

```python
Function | Notation | Growth Rate

Constant | ≈ 1 | Doesn't grow 
Logarithmic | ≈ log n | Very slow growth 
Linear | ≈ n | Grows steadily
N-Log-N | ≈ n log n | Moderate growth 
Quadratic | ≈ n^2 | Fast growth 
Cubic | ≈ n^3 | Very fast growth 
Exponential | ≈ 2^n | Explosive growth 


Example - Sorting 1 million items:
- Insertion sort ((n^2)/4) : ~70 hours
- Merge sort (2n log n) : ~40 seconds
```

**3.6 Primitive Operations**

```python
Definition : Basic computations taking constant time

Examples : 
- Evaluation an expression
- Assigning a value to a variable
- Indexing into an array
- Calling a method
- Returning from a method
```

```python
def find_max(data):
    biggest = data[0]      # Assignment + indexing (2 ops)
    for val in data:       # Loop header
        if val > biggest:  # Comparison
            biggest = val  # Assignment
    return biggest         # Return
```

**3.7 Big-Oh Notation (O)**

```python
Definition : f(n) is O(g(n)) if there exist constants c > 0
and n_0 >= 1 such that :

f(n) <= c * g(n) for all n >= n_0
```

```python
# Example: 2n + 10 is O(n)
# Need: 2n + 10 ≤ c·n for n ≥ n₀
# Pick c = 3, n₀ = 10
# Check: 2(10) + 10 = 30 ≤ 3(10) = 30 ✓

# More examples:
# 7n - 2 is O(n)        (c = 7, n₀ = 1)
# 3n³ + 20n² + 5 is O(n³) (c = 4, n₀ = 21)
# 3 log n + 5 is O(log n)  (c = 8, n₀ = 2)
```

```python
Intuition : Big-Oh means "asymptotically less than or equal to"
```

**3.8 Computing Prefix Averages (Example)**

```python
Problem : For array X, compute A where A[i] = average of 
X[0] to X[i]
```

```python
# Version 1 - O(n²) - SLOW
def prefix_average1(S):
    n = len(S)
    A = [0] * n
    for j in range(n):
        total = 0
        for i in range(j + 1):  # Nested loop!
            total += S[i]
        A[j] = total / (j + 1)
    return A

# Version 2 - Still O(n²) - sum() hides but doesn't help
def prefix_average2(S):
    n = len(S)
    A = [0] * n
    for j in range(n):
        A[j] = sum(S[0:j+1]) / (j+1)  # sum is still O(n)
    return A

# Version 3 - O(n) - FAST!
def prefix_average3(S):
    n = len(S)
    A = [0] * n
    total = 0
    for j in range(n):
        total += S[j]      # Keep running sum - O(1) per iteration
        A[j] = total / (j + 1)
    return A
```

```python
Why version 3 is faster :
- Version 1 : 1 + 2 + 3 + ... + n = n(n+1)/2 = O(n^2)
- Version 3 : Just n operations = O(n)
```
