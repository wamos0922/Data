```python
"""
Queue Example - Section 6.3 Demonstration
Shows the FIFO (First-In, First-Out) behavior of a queue.
"""

class EmptyQueueException(Exception):
    """Custom exception for empty queue operations."""
    pass


class ArrayQueue:
    """FIFO queue implementation using a Python list as underlying storage."""
    
    DEFAULT_CAPACITY = 10
    
    def __init__(self):
        """Create an empty queue."""
        self._data = [None] * ArrayQueue.DEFAULT_CAPACITY
        self._size = 0
        self._front = 0
    
    def __len__(self):
        """Return the number of elements in the queue."""
        return self._size
    
    def is_empty(self):
        """Return True if the queue is empty."""
        return self._size == 0
    
    def first(self):
        """Return (but do not remove) the element at the front of the queue."""
        if self.is_empty():
            raise EmptyQueueException("Queue is empty")
        return self._data[self._front]
    
    def enqueue(self, e):
        """Add an element to the back of queue."""
        if self._size == len(self._data):
            self._resize(2 * len(self._data))
        
        avail = (self._front + self._size) % len(self._data)
        self._data[avail] = e
        self._size += 1
    
    def dequeue(self):
        """Remove and return the element at the front of the queue."""
        if self.is_empty():
            raise EmptyQueueException("Queue is empty")
        
        answer = self._data[self._front]
        self._data[self._front] = None  # Help garbage collection
        self._front = (self._front + 1) % len(self._data)
        self._size -= 1
        return answer
    
    def _resize(self, cap):
        """Resize to a new list of capacity >= len(self)."""
        old = self._data
        self._data = [None] * cap
        walk = self._front
        for k in range(self._size):
            self._data[k] = old[walk]
            walk = (1 + walk) % len(old)
        self._front = 0
    
    def display(self):
        """Helper method to show current queue state (for learning)."""
        if self.is_empty():
            print("[]")
            return
        
        # Collect elements in order from front to rear
        elements = []
        pos = self._front
        for _ in range(self._size):
            elements.append(repr(self._data[pos]))
            pos = (pos + 1) % len(self._data)
        
        print(f"[{', '.join(elements)}]")


# ============================================================
# SECTION 6.3 EXAMPLE - Exactly as shown in the PDF
# ============================================================

def run_section_63_example():
    """Run the exact example from Section 6.3 of the PDF."""
    
    print("=" * 60)
    print("Section 6.3 Queue Example")
    print("=" * 60)
    print()
    
    Q = ArrayQueue()
    
    # Format: Operation | Return Value | Queue State
    print(f"{'Operation':<20} | {'Return Value':<12} | Queue State (front ← → rear)")
    print("-" * 60)
    
    # Q.enqueue(5)
    Q.enqueue(5)
    print(f"{'Q.enqueue(5)':<20} | {'-':<12} | ", end="")
    Q.display()
    
    # Q.enqueue(3)
    Q.enqueue(3)
    print(f"{'Q.enqueue(3)':<20} | {'-':<12} | ", end="")
    Q.display()
    
    # len(Q)
    length = len(Q)
    print(f"{'len(Q)':<20} | {length:<12} | ", end="")
    Q.display()
    
    # Q.dequeue()
    dequeued = Q.dequeue()
    print(f"{'Q.dequeue()':<20} | {dequeued:<12} | ", end="")
    Q.display()
    
    # Q.is_empty()
    is_empty = Q.is_empty()
    print(f"{'Q.is_empty()':<20} | {is_empty:<12} | ", end="")
    Q.display()
    
    # Q.dequeue()
    dequeued = Q.dequeue()
    print(f"{'Q.dequeue()':<20} | {dequeued:<12} | ", end="")
    Q.display()
    
    # Q.is_empty()
    is_empty = Q.is_empty()
    print(f"{'Q.is_empty()':<20} | {is_empty:<12} | ", end="")
    Q.display()
    
    # Q.dequeue() - This will raise an exception
    print(f"{'Q.dequeue()':<20} | {'error':<12} | ", end="")
    try:
        Q.dequeue()
        print("This should not print!")
    except EmptyQueueException as e:
        print(f"Exception: {e}")
    
    # Q.enqueue(7)
    Q.enqueue(7)
    print(f"{'Q.enqueue(7)':<20} | {'-':<12} | ", end="")
    Q.display()
    
    # Q.enqueue(9)
    Q.enqueue(9)
    print(f"{'Q.enqueue(9)':<20} | {'-':<12} | ", end="")
    Q.display()
    
    # Q.first()
    first = Q.first()
    print(f"{'Q.first()':<20} | {first:<12} | ", end="")
    Q.display()
    
    # Q.enqueue(4)
    Q.enqueue(4)
    print(f"{'Q.enqueue(4)':<20} | {'-':<12} | ", end="")
    Q.display()
    
    # len(Q)
    length = len(Q)
    print(f"{'len(Q)':<20} | {length:<12} | ", end="")
    Q.display()
    
    # Q.dequeue()
    dequeued = Q.dequeue()
    print(f"{'Q.dequeue()':<20} | {dequeued:<12} | ", end="")
    Q.display()


# ============================================================
# ADDITIONAL DEMONSTRATIONS FOR BETTER UNDERSTANDING
# ============================================================

def demonstrate_fifo_behavior():
    """Demonstrate FIFO behavior with more elements."""
    print("\n" + "=" * 60)
    print("FIFO Behavior Demonstration")
    print("=" * 60)
    
    q = ArrayQueue()
    
    print("\nEnqueuing: 10, 20, 30, 40, 50")
    for num in [10, 20, 30, 40, 50]:
        q.enqueue(num)
    
    print("Queue state: ", end="")
    q.display()
    
    print("\nDequeuing all elements (FIFO order):")
    while not q.is_empty():
        print(f"  Dequeued: {q.dequeue()}")
    
    print("\nFinal queue state: ", end="")
    q.display()


def demonstrate_circular_behavior():
    """Show how the circular array wraps around."""
    print("\n" + "=" * 60)
    print("Circular Array Behavior Demonstration")
    print("=" * 60)
    
    # Create a small custom queue that shows internal indices
    class ObservableQueue(ArrayQueue):
        DEFAULT_CAPACITY = 4  # Small capacity to show wrap-around quickly
        
        def show_internal(self):
            print(f"  Internal array: {self._data}")
            print(f"  front index: {self._front}, size: {self._size}")
            print(f"  rear index (available): {(self._front + self._size) % len(self._data)}")
    
    q = ObservableQueue()
    
    print("\nInitial empty queue:")
    q.show_internal()
    
    print("\nEnqueue A, B, C:")
    for letter in ['A', 'B', 'C']:
        q.enqueue(letter)
        print(f"  After enqueue('{letter}'):")
        q.show_internal()
    
    print("\nDequeue once:")
    removed = q.dequeue()
    print(f"  Removed: {removed}")
    q.show_internal()
    
    print("\nEnqueue D, E (shows wrap-around):")
    q.enqueue('D')
    q.show_internal()
    q.enqueue('E')
    q.show_internal()
    
    print("\nDequeue all:")
    while not q.is_empty():
        print(f"  Dequeued: {q.dequeue()}")
        q.show_internal()


# ============================================================
# RUN ALL DEMONSTRATIONS
# ============================================================

if __name__ == "__main__":
    # Run the main Section 6.3 example
    run_section_63_example()
    
    # Run additional demonstrations
    demonstrate_fifo_behavior()
    demonstrate_circular_behavior()
```

```python
def _resize(self, cap):
    """Resize to a new list of capacity >= len(self)."""
    old = self._data                    # Keep track of existing list
    self._data = [None] * cap           # Allocate new list with larger capacity
    walk = self._front
    
    for k in range(self._size):         # Only consider existing elements
        self._data[k] = old[walk]       # Shift indices to start at 0
        walk = (1 + walk) % len(old)    # Use old size as modulus
    
    self._front = 0                     # Front has been realigned to index 0
```

```python
def is_matched(expr):
    """Return True if all delimiters are properly matched; False otherwise."""
    lefty = "({["      # Opening delimiters
    righty = ")}]"     # Respective closing delimiters
    
    S = ArrayStack()
    
    for c in expr:
        if c in lefty:
            S.push(c)                      # Push left delimiter on stack
        elif c in righty:
            if S.is_empty():
                return False               # Nothing to match with
            if righty.index(c) != lefty.index(S.pop()):
                return False               # Mismatched type
    
    return S.is_empty()                    # Were all symbols matched?
```

```python
def is_matched_html(raw):
    """Return True if all HTML tags are properly matched; False otherwise."""
    S = ArrayStack()
    j = raw.find('<')                     # Find first '<' character
    
    while j != -1:
        k = raw.find('>', j + 1)          # Find next '>' character
        if k == -1:
            return False                  # Invalid tag
        tag = raw[j + 1:k]                # Strip away < >
        
        if not tag.startswith('/'):       # Opening tag
            S.push(tag)
        else:                             # Closing tag
            if S.is_empty():
                return False              # Nothing to match with
            if tag[1:] != S.pop():
                return False              # Mismatched tag
        
        j = raw.find('<', k + 1)          # Find next '<' character
    
    return S.is_empty()                   # Were all opening tags matched?
```

*6.10.2 Two-Stack Algorithm*

```python
def do_op(valStk, opStk):
    """Perform one operation: pop two values and one operator, push result."""
    x = valStk.pop()
    y = valStk.pop()
    op = opStk.pop()
    
    if op == '+':
        valStk.push(y + x)
    elif op == '-':
        valStk.push(y - x)
    elif op == '*':
        valStk.push(y * x)
    elif op == '/':
        valStk.push(y / x)

def repeat_ops(valStk, opStk, ref_op, precedence):
    """Pop and execute operators with precedence >= ref_op."""
    while len(valStk) > 1 and precedence(ref_op) <= precedence(opStk.top()):
        do_op(valStk, opStk)
```

*6.10.3 Evaluation Algorithm*

*6.10.4 Example : 14 - 3 * 2 + 7*

```python
"""
Expression Evaluation - Section 6.10.4 Example
Evaluates arithmetic expressions using two stacks (operator and value stacks).

Example: 14 - 3 * 2 + 7 = 15
"""

class EmptyStackException(Exception):
    """Custom exception for empty stack operations."""
    pass


class ArrayStack:
    """LIFO Stack implementation using a Python list as underlying storage."""
    
    def __init__(self):
        """Create an empty stack."""
        self._data = []
    
    def __len__(self):
        """Return the number of elements in the stack."""
        return len(self._data)
    
    def is_empty(self):
        """Return True if the stack is empty."""
        return len(self._data) == 0
    
    def push(self, e):
        """Add element e to the top of the stack."""
        self._data.append(e)
    
    def top(self):
        """Return (but do not remove) the element at the top of the stack."""
        if self.is_empty():
            raise EmptyStackException("Stack is empty")
        return self._data[-1]
    
    def pop(self):
        """Remove and return the element from the top of the stack (LIFO)."""
        if self.is_empty():
            raise EmptyStackException("Stack is empty")
        return self._data.pop()
    
    def display(self):
        """Helper method to show current stack state."""
        if self.is_empty():
            return "[]"
        return f"[{', '.join(str(elem) for elem in self._data)}]"


class ExpressionEvaluator:
    """
    Evaluates arithmetic expressions using two-stack algorithm.
    Supports +, -, *, / operators with proper precedence.
    """
    
    def __init__(self):
        """Initialize the evaluator with operator precedence rules."""
        self.precedence = {
            '$': 0,  # End-of-input marker (lowest precedence)
            '+': 1,
            '-': 1,
            '*': 2,
            '/': 2
        }
    
    def _apply_operator(self, val_stk, op_stk):
        """
        Apply one operator: pop two values and one operator,
        compute result, push back to value stack.
        """
        # Pop operands (note: right operand is popped first)
        right = val_stk.pop()
        left = val_stk.pop()
        op = op_stk.pop()
        
        # Perform operation
        if op == '+':
            result = left + right
        elif op == '-':
            result = left - right
        elif op == '*':
            result = left * right
        elif op == '/':
            if right == 0:
                raise ZeroDivisionError("Division by zero")
            result = left / right
        else:
            raise ValueError(f"Unknown operator: {op}")
        
        # Push result back
        val_stk.push(result)
        return result, op, left, right
    
    def _should_pop_operator(self, current_op, top_op):
        """Check if we should pop the top operator (higher or equal precedence)."""
        return self.precedence[current_op] <= self.precedence[top_op]
    
    def evaluate(self, expression):
        """
        Evaluate an arithmetic expression.
        
        Args:
            expression: String like "14 - 3 * 2 + 7"
        
        Returns:
            Numerical result of the expression
        """
        # Initialize stacks
        val_stk = ArrayStack()  # Value stack
        op_stk = ArrayStack()   # Operator stack
        
        # Tokenize the expression (split by spaces for simplicity)
        # For a more robust version, we'd parse properly, but this works for our example
        tokens = expression.split()
        
        print("\n" + "=" * 80)
        print(f"Evaluating: {expression}")
        print("=" * 80)
        print(f"{'Step':<4} | {'Token':<6} | {'Action':<30} | {'Value Stack':<20} | {'Op Stack':<15}")
        print("-" * 80)
        
        step = 1
        
        for token in tokens:
            # Check if token is a number
            if self._is_number(token):
                val_stk.push(float(token))
                print(f"{step:<4} | {token:<6} | {'Push value':<30} | {val_stk.display():<20} | {op_stk.display():<15}")
            
            # Check if token is an operator
            elif token in self.precedence:
                # Pop operators with higher or equal precedence
                while (len(op_stk) > 0 and 
                       op_stk.top() != '(' and 
                       self._should_pop_operator(token, op_stk.top())):
                    
                    result, op, left, right = self._apply_operator(val_stk, op_stk)
                    print(f"{step:<4} | {token:<6} | {'Pop & apply ' + op:>30} | {val_stk.display():<20} | {op_stk.display():<15}")
                    step += 1
                
                # Push current operator
                op_stk.push(token)
                print(f"{step:<4} | {token:<6} | {'Push operator':<30} | {val_stk.display():<20} | {op_stk.display():<15}")
            
            # Handle parentheses (for more complex expressions)
            elif token == '(':
                op_stk.push(token)
                print(f"{step:<4} | {token:<6} | {'Push (':<30} | {val_stk.display():<20} | {op_stk.display():<15}")
            
            elif token == ')':
                while not op_stk.is_empty() and op_stk.top() != '(':
                    result, op, left, right = self._apply_operator(val_stk, op_stk)
                    print(f"{step:<4} | {token:<6} | {'Pop & apply ' + op:>30} | {val_stk.display():<20} | {op_stk.display():<15}")
                    step += 1
                if not op_stk.is_empty() and op_stk.top() == '(':
                    op_stk.pop()  # Remove '('
                    print(f"{step:<4} | {token:<6} | {'Pop (':<30} | {val_stk.display():<20} | {op_stk.display():<15}")
            
            step += 1
        
        # Process remaining operators
        while len(op_stk) > 0:
            result, op, left, right = self._apply_operator(val_stk, op_stk)
            print(f"{step:<4} | {'END':<6} | {'Pop & apply ' + op:>30} | {val_stk.display():<20} | {op_stk.display():<15}")
            step += 1
        
        # Final result
        final_result = val_stk.top()
        print("-" * 80)
        print(f"\n✅ Final Result: {expression} = {final_result}")
        
        return final_result
    
    def _is_number(self, token):
        """Check if a token is a number (integer or float)."""
        try:
            float(token)
            return True
        except ValueError:
            return False
    
    def evaluate_with_trace(self, expression, show_detailed=True):
        """
        Evaluate expression with detailed step-by-step tracing.
        This is specifically for the 14 - 3 * 2 + 7 example.
        """
        # Tokenize
        tokens = expression.split()
        
        # Initialize
        val_stk = ArrayStack()
        op_stk = ArrayStack()
        
        print("\n" + "=" * 70)
        print(f"EXPRESSION EVALUATION: {expression}")
        print("=" * 70)
        print("\nAlgorithm: Two-Stack Evaluation")
        print("-" * 70)
        
        for token in tokens:
            if self._is_number(token):
                val_stk.push(float(token))
                print(f"\n📥 Read number: {token}")
                print(f"   Value Stack: {val_stk.display()}")
                print(f"   Op Stack:    {op_stk.display()}")
            
            elif token in self.precedence:
                print(f"\n📌 Read operator: {token}")
                
                # Pop higher/equal precedence operators
                while (len(op_stk) > 0 and 
                       op_stk.top() != '(' and 
                       self.precedence[token] <= self.precedence[op_stk.top()]):
                    
                    right = val_stk.pop()
                    left = val_stk.pop()
                    op = op_stk.pop()
                    
                    print(f"   🔄 Popping {op} (precedence {self.precedence[op]})")
                    print(f"      Applying: {left} {op} {right}")
                    
                    if op == '+':
                        result = left + right
                    elif op == '-':
                        result = left - right
                    elif op == '*':
                        result = left * right
                    elif op == '/':
                        result = left / right
                    
                    val_stk.push(result)
                    print(f"      Result: {result}")
                    print(f"   Value Stack: {val_stk.display()}")
                    print(f"   Op Stack:    {op_stk.display()}")
                
                # Push current operator
                op_stk.push(token)
                print(f"   ✅ Pushing {token} (precedence {self.precedence[token]})")
                print(f"   Value Stack: {val_stk.display()}")
                print(f"   Op Stack:    {op_stk.display()}")
        
        # Process remaining operators
        print(f"\n🔚 End of expression. Processing remaining operators...")
        while len(op_stk) > 0:
            right = val_stk.pop()
            left = val_stk.pop()
            op = op_stk.pop()
            
            print(f"\n   Applying: {left} {op} {right}")
            
            if op == '+':
                result = left + right
            elif op == '-':
                result = left - right
            elif op == '*':
                result = left * right
            elif op == '/':
                result = left / right
            
            val_stk.push(result)
            print(f"   Result: {result}")
            print(f"   Value Stack: {val_stk.display()}")
            print(f"   Op Stack:    {op_stk.display()}")
        
        final_result = val_stk.top()
        print("\n" + "=" * 70)
        print(f"✅ FINAL RESULT: {expression} = {final_result}")
        print("=" * 70)
        
        return final_result


# ============================================================
# SECTION 6.10.4 SPECIFIC EXAMPLE: 14 - 3 * 2 + 7
# ============================================================

def run_section_6104_example():
    """Run the exact example from Section 6.10.4 of the PDF."""
    
    print("\n" + "=" * 80)
    print("SECTION 6.10.4: Expression Evaluation Example")
    print("Expression: 14 - 3 * 2 + 7")
    print("Expected result: 15 (since 14 - (3*2) + 7 = 14 - 6 + 7 = 15)")
    print("=" * 80)
    
    evaluator = ExpressionEvaluator()
    
    # Evaluate with detailed tracing
    result = evaluator.evaluate_with_trace("14 - 3 * 2 + 7")
    
    return result


# ============================================================
# ADDITIONAL EXAMPLES FOR PRACTICE
# ============================================================

def run_additional_examples():
    """Run additional expression examples to demonstrate the algorithm."""
    
    print("\n\n" + "=" * 80)
    print("ADDITIONAL EXPRESSION EXAMPLES")
    print("=" * 80)
    
    evaluator = ExpressionEvaluator()
    
    examples = [
        ("3 + 4 * 2", 3 + 4 * 2),                    # 3 + 8 = 11
        ("10 - 2 * 3 + 4", 10 - 2 * 3 + 4),          # 10 - 6 + 4 = 8
        ("2 * 3 + 4 * 5", 2 * 3 + 4 * 5),            # 6 + 20 = 26
        ("20 / 4 + 3", 20 / 4 + 3),                  # 5 + 3 = 8
        ("5 + 3 * 2 - 8 / 4", 5 + 3 * 2 - 8 / 4),    # 5 + 6 - 2 = 9
    ]
    
    for expr, expected in examples:
        print(f"\n{'=' * 60}")
        print(f"Expression: {expr}")
        print(f"Expected:   {expected}")
        
        # Simple evaluation without full trace
        val_stk = ArrayStack()
        op_stk = ArrayStack()
        prec = {'+': 1, '-': 1, '*': 2, '/': 2}
        
        tokens = expr.split()
        
        for token in tokens:
            if token.replace('.', '').isdigit() or (token[0] == '-' and token[1:].replace('.', '').isdigit()):
                val_stk.push(float(token))
            elif token in prec:
                while len(op_stk) > 0 and prec[token] <= prec[op_stk.top()]:
                    right = val_stk.pop()
                    left = val_stk.pop()
                    op = op_stk.pop()
                    if op == '+': val_stk.push(left + right)
                    elif op == '-': val_stk.push(left - right)
                    elif op == '*': val_stk.push(left * right)
                    elif op == '/': val_stk.push(left / right)
                op_stk.push(token)
        
        while len(op_stk) > 0:
            right = val_stk.pop()
            left = val_stk.pop()
            op = op_stk.pop()
            if op == '+': val_stk.push(left + right)
            elif op == '-': val_stk.push(left - right)
            elif op == '*': val_stk.push(left * right)
            elif op == '/': val_stk.push(left / right)
        
        result = val_stk.top()
        print(f"Result:     {result}")
        print(f"Match:      {'✅' if result == expected else '❌'}")


# ============================================================
# STEP-BY-STEP MANUAL TRACE OF 14 - 3 * 2 + 7
# ============================================================

def manual_trace():
    """Show manual step-by-step trace matching the PDF exactly."""
    
    print("\n" + "=" * 80)
    print("MANUAL STEP-BY-STEP TRACE (Matching PDF Section 6.10.4)")
    print("=" * 80)
    
    print("""
Expression: 14 - 3 * 2 + 7
Precedence: *, / have precedence 2; +, - have precedence 1

Step-by-step execution:

Step 1: Read '14' (number)
        Value Stack: [14]
        Operator Stack: []

Step 2: Read '-' (operator)
        Stack empty → push '-'
        Value Stack: [14]
        Operator Stack: [-]

Step 3: Read '3' (number)
        Push 3
        Value Stack: [14, 3]
        Operator Stack: [-]

Step 4: Read '*' (operator)
        Compare precedence: '*' (2) > '-' (1) → don't pop, just push
        Value Stack: [14, 3]
        Operator Stack: [-, *]

Step 5: Read '2' (number)
        Push 2
        Value Stack: [14, 3, 2]
        Operator Stack: [-, *]

Step 6: Read '+' (operator)
        Compare precedence: '+' (1) ≤ '*' (2) → pop '*'
        Pop '*', pop 2 and 3 → compute 3 * 2 = 6, push 6
        Value Stack: [14, 6]
        Operator Stack: [-]
        
        Compare again: '+' (1) ≤ '-' (1) → pop '-'
        Pop '-', pop 6 and 14 → compute 14 - 6 = 8, push 8
        Value Stack: [8]
        Operator Stack: []
        
        Push '+'
        Value Stack: [8]
        Operator Stack: [+]

Step 7: Read '7' (number)
        Push 7
        Value Stack: [8, 7]
        Operator Stack: [+]

Step 8: End of expression
        Pop '+', pop 7 and 8 → compute 8 + 7 = 15, push 15
        Value Stack: [15]
        Operator Stack: []

Final Result: 15
    """)


# ============================================================
# RUN ALL DEMONSTRATIONS
# ============================================================

if __name__ == "__main__":
    # Run the main Section 6.10.4 example
    run_section_6104_example()
    
    # Show manual trace
    manual_trace()
    
    # Run additional examples
    run_additional_examples()
```

*6.11.2 Quadratic Algorithm (Slow)*

```python
def spans_quadratic(X):
    """Compute spans - O(n²) time."""
    n = len(X)
    S = [0] * n
    
    for i in range(n):
        s = 1
        while s <= i and X[i - s] <= X[i]:
            s += 1
        S[i] = s
    
    return S
```

*6.11.3 Linear Algorithm Using Stack (fast)*

```python
def spans_linear(X):
    """Compute spans - O(n) time using stack."""
    n = len(X)
    S = [0] * n
    stack = ArrayStack()
    
    for i in range(n):
        # Pop indices of elements <= X[i]
        while not stack.is_empty() and X[stack.top()] <= X[i]:
            stack.pop()
        
        # Compute span
        if stack.is_empty():
            S[i] = i + 1          # All previous elements are ≤ X[i]
        else:
            S[i] = i - stack.top() # Distance to next larger element
        
        # Push current index
        stack.push(i)
    
    return S
```
