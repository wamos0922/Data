**4.1 What is Recursion ?**

```python
Definition : When a method calls itself

Classic Example : Factorial Function
```

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
```

```python
Key Insight : The problem breaks down into
              smaller versions of itself
```

**4.2 Two Essential Parts of Every Recursive Method**

*4.2.1 Base Case(s)*

```python
- Values where no recursive call is made
- Must have at least one base case
- Every recursive chain must eventually reach a base case
```

*4.2.1 Recursive Call(s)*

```python
- Calls to the same method
- Each call must make progress toward a base case
- Progress means reducing the problem size
```

```python
Visual Formula :

Recursive Method = Base Case(s) + Recursive Call(s) that make progress
```

**4.3 Visualizing Recursion : The Recursion Trace**

```python
Components of a recursion trace :

- A box for each recursive call
- An arrow from caller -> callee
- An arrow from calle -> caller (showing return value)
```

```python
Example trace for factorial(4) : 

factorial(4) → 24
    │
    └─→ factorial(3) → 6
            │
            └─→ factorial(2) → 2
                    │
                    └─→ factorial(1) → 1
                            │
                            └─→ factorial(0) → 1
```

**4.4 Example : English Ruler**

```python
Problem : Print ticks and number like a ruler.

Recursive Definition :
An internal with central tick length L >=1 consists of :

1. An interval with central tick length L - 1
2. A single tick of length L
3. An interval with central tick length L - 1
```

```python
Python implementation :
```

```python
def draw_line(tick_length, tick_label=''):
    line = '-' * tick_length
    if tick_label:
        line += ' ' + tick_label
    print(line)

def draw_interval(center_length):
    if center_length > 0:
        draw_interval(center_length - 1)   # Top half
        draw_line(center_length)            # Middle tick
        draw_interval(center_length - 1)   # Bottom half

def draw_ruler(num_inches, major_length):
    draw_line(major_length, '0')
    for j in range(1, num_inches + 1):
        draw_interval(major_length - 1)
        draw_line(major_length, str(j))
```

**4.5 Binary Search (Linear Recursion Example)**

```python
Problem : Find target in a sorted array.

Algorithm :
```

```python
def binary_search(data, target, low, high):
    if low > high:                      # Base: empty range
        return False
    
    mid = (low + high) // 2
    
    if target == data[mid]:             # Base: found
        return True
    elif target < data[mid]:            # Recur on left half
        return binary_search(data, target, low, mid - 1)
    else:                               # Recur on right half
        return binary_search(data, target, mid + 1, high)
```

```python
Three cases :

1. target == data[mid] -> Found! (base case)
2. target < data[mid] -> Search left half
3. target > data[mid] -> Search right half

Time complexity : O(log n)
```

**4.6 Types of Recursion**

*4.6.1 Linear Recursion*

```python
Definition : Makes at most one recursive call per invocation.

Characteristics :

- Tests for base cases first
- Makes single recursice call
- Progresses linearly toward base case

Example : Array reversal
```

```python
def reverse(S, start, stop):
    if start < stop - 1:                # At least 2 elements
        S[start], S[stop - 1] = S[stop - 1], S[start]  # Swap
        reverse(S, start + 1, stop - 1) # Single recursive call
```

```python
Example : Linear sum
```

```python
def linear_sum(A, n):
    if n == 1:
        return A[0]
    else:
        return linear_sum(A, n - 1) + A[n - 1]
```

*4.6.2 Binary Recursion*

```python
Definition : Makes two recursive calls per invocation.


Example : Drawing ruler ticks
```

```python
def draw_interval(center_length):
    if center_length > 0:
        draw_interval(center_length - 1)   # Call 1
        draw_line(center_length)
        draw_interval(center_length - 1)   # Call 2
```

```python
Example : Binary sum of array
```

```python
def binary_sum(A, i, n):
    if n == 1:
        return A[i]
    else:
        return binary_sum(A, i, n // 2) + binary_sum(A, i + n // 2, n // 2)
```

```python
Usage : binary_sum(A, 0, len(A))
```

*4.6.3 Multiple Recursion*

```python
Definition : Makes many recursive calls (more that two)
             per invocation.


Ecample :

- Summation puzzles (POT + PAN = BIB)
- Sudoku solvers
- Tree traversals (each node spawns calls to all children)
```

**4.7 Computing Powers Efficiently**

*4.7.1 Simple Linear Power(O(n) time)*

```python
def power_slow(x, n):
    if n == 0:
        return 1
    else:
        return x * power_slow(x, n - 1)   # n recursive calls
```

```python
Problem : Takes n steps -> inefficient for large n
```

*4.7.2 Recursive Squaring (O(log n) time)*

```python
Mathematical definition :

p(x, n) = 1                          if n = 0
p(x, n) = x × p(x, (n-1)/2)²         if n > 0 and n is odd
p(x, n) = p(x, n/2)²                 if n > 0 and n is even
```

```python
Python implementation :
```

```python
def power_fast(x, n):
    if n == 0:
        return 1
    elif n % 2 == 0:                  # Even
        return power_fast(x, n // 2) ** 2
    else:                              # Odd
        return x * power_fast(x, (n - 1) // 2) ** 2
```

**4.8 Fibonacci Numbers : Bad vs Good Recursion**

*4.8.1 Definition*

```python
F₀ = 0
F₁ = 1
Fᵢ = Fᵢ₋₁ + Fᵢ₋₂    for i > 1
```

*4.8.2 Bad Binary Recursion (Exponential Time)*

```python
def bad_fib(k):
    if k <= 1:
        return k
    return bad_fib(k - 1) + bad_fib(k - 2)   # Two recursive calls!
```

![AltText](4.8.2.png)

Problem : Recalculates same Fibonacci numbers many times.

*4.8.3 Good Linear Recursion*

```python
def good_fib(k):
    """Returns (F_k, F_{k-1}) as a tuple"""
    if k == 1:
        return (1, 0)
    else:
        i, j = good_fib(k - 1)        # Single recursive call
        return (i + j, i)
```

```python
How it works :

- Returns pair of consecutive Fibonacci numbers
- Each value computed exactly once
- Only k-1 recursive calls total

Example trace for good_fib(5) : 

good_fib(5) → (5, 3)
    ↑ from good_fib(4) → (3, 2)
        ↑ from good_fib(3) → (2, 1)
            ↑ from good_fib(2) → (1, 1)
                ↑ from good_fib(1) → (1, 0)f
```

```python
Example : Array reversal with extra parameters (good design)
```

```python
def good_reverse(S, start, stop):   # Extra parameters
    if start < stop - 1:
        S[start], S[stop - 1] = S[stop - 1], S[start]
        good_reverse(S, start + 1, stop - 1)

# Wrapper function for user convenience
def reverse(S):
    good_reverse(S, 0, len(S))
```

```python
Design Patterns :

- Public wrapper : Takes simple parameters (user-friendly)
- Private recursive helper : Takes extra parameters (track
                             recursion state)
```

**4.12 When to Use Recursion**

```python
Use Recursion When :

- Problem has natural recursive structure
- Code clarity is more important than performance
- Problem is inherently hierarchical
- Using divide-and-conquer


Use Iteration instead when :

- Problem is simple linear repetition
- Performance is critical
- Deep recursion risk
- Memory usage is a concern
```

**4.13 Quick Recap Checklist**

```python
- Every recursive method has base case(s) and recursive call(s)
- Recursive calls must make progress toward base case
- Linear recursion = 1 recursive call (simplest)
- Binary recursion = 2 recursive calls
- Multiple recursion = many recursive calls 
- Recursive squaring computes powers in O(log n)
- Good Fibonacci uses linear recursion, not Binary
- Sometimes need extra parameters for recursion
- Recursion trace = boxes + arrows showing call flow
```

**4.14 Key Formulas to Remember**

```python
Factorial:      n! = n × (n-1)!
Power (fast):   xⁿ = (xⁿᐟ²)²              if n even
                xⁿ = x × (x⁽ⁿ⁻¹⁾ᐟ²)²      if n odd
Fibonacci:      F₀ = 0, F₁ = 1, Fᵢ = Fᵢ₋₁ + Fᵢ₋₂
Binary Search:  T(n) = T(n/2) + O(1) → O(log n)
```
