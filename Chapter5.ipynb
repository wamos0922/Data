**5.4 Dynamic Array Operations**

*5.4.1 Insertion Operation*

```python
Problem : Insert element at position i 
          (shifts elements to the right)


Process :

1. Make room by shifting elements A[i] through A[n-1] 
   one position forward
2. Insert new element at position i



Visual :

Before: [a, b, c, d, e]  insert 'x' at index 2
After:  [a, b, x, c, d, e]
              ↑
         Shifted c,d,e right


Time complexity :

- Worst case (i=0): Shift n elements -> O(n)
- Best case (i = n) : No shift -> O(1)
```

*5.4.2 Removal Operation remove(i)*

```python
Problem : Remove element at position i (shifts elements to fill the hole)

Process : 

1. Remove element at index i
2. Shift elements A[i+1] through A[n-1] one position backward



Visual :

Before: [a, b, x, c, d]  remove index 2 (the 'x')
After:  [a, b, c, d]
              ↓
         Shifted c,d left



Time complexity :

- Worst case (i = 0) : Shift n-1 elements -> O(n)
- Best case (i = n-1) : No shift -> O(1)
```

**5.6 Growable Array (Dynamic Array)**

```python
Problem : Fixed-size arrays cannot grow beyond their initial capacity.

Solution : When array is full, replace it with a larger array.


Algorithm add(o) (append to end):

if n == capacity - 1:           # Array is full
    Create new array A' of larger size
    Copy all elements from S to A'
    S ← A'                       # Replace old array
n ← n + 1
S[n - 1] ← o                     # Add new element
```

**5.7 Amortized Analysis**

```python
Definition : The average time per operation over a sequence of operations.

Formula : Amortized time = T(n) / n where T(n) = total time for n operations

Why it matters : Even if some operations are expensive (resizing),
                 the average can still be good.
```

*5.7.1 Incremental Strategy Analysis (Bad)*

```python
Assumptions :

- Start with array of size 1
- Increase by constant c each time
- k = n / c , resizes needed


Total time T(n) for n append operations :

T(n) = n + c + 2c + 3c + 4c + ... + kc
     = n + c(1 + 2 + 3 + ... + k)
     = n + c × k(k + 1)/2

Result : T(n) is O(n²) -> Amortized time = O(n)


Conclusion : Incremental strategy is inefficient for large n.
```

*5.7.1 Doubling Strategy Analysis (Good)*

```python
Assumptions :

- Start with array of size 1
- Double size each time : 1, 2, 4, 8, 16, ...
- k = log_2 n , resizes needed


Total time T(n) for n append operations :

T(n) = n + 1 + 2 + 4 + 8 + ... + 2^k
     = n + (2^(k+1) - 1)
     = n + 2n - 1
     = 3n - 1

Result : T(n) is O(n) -> Amortized time = O(1)

Conclusion : Doubling strategy is highly efficient (constant average time).
```

**5.8 Python Implementation of DynamicArray**

```python
import ctypes  # Provides low-level arrays (like C arrays)

class DynamicArray:
    """A dynamic array class akin to a simplified Python list."""
    
    def __init__(self):
        """Create an empty array."""
        self.n = 0                    # Number of elements
        self.capacity = 1             # Current array capacity
        self.A = self._make_array(self.capacity)  # Low-level array
    
    def __len__(self):
        """Return number of elements stored."""
        return self.n
    
    def __getitem__(self, k):
        """Return element at index k (supports A[k] syntax)."""
        if not (0 <= k < self.n):
            raise IndexError('invalid index')
        return self.A[k]              # Retrieve from array
    
    def append(self, obj):
        """Add object to end of the array."""
        if self.n == self.capacity:   # Array is full
            self._resize(2 * self.capacity)  # DOUBLE the capacity!
        self.A[self.n] = obj          # Add new element
        self.n += 1
    
    def _resize(self, c):
        """Resize internal array to capacity c."""
        B = self._make_array(c)       # Create new larger array
        for k in range(self.n):       # Copy all elements
            B[k] = self.A[k]
        self.A = B                    # Replace old array
        self.capacity = c
    
    def _make_array(self, c):
        """Return new array with capacity c (using ctypes)."""
        return (c * ctypes.py_object)()  # Array of Python object references
```
